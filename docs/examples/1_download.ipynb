# Download the full dataset

Use `plinder_download` when you need a local copy, for example on a compute node without internet access.

## Choose a local directory

Files are cached in `~/.local/share/plinder/<release-month>` by default. Set `PLINDER_MOUNT` before importing PLINDER to choose another location. Inspect the configured directory with:

````python
from plinder.core import get_config

cfg = get_config()
print(cfg.data.plinder_dir)
````

## Download release files

`plinder_download` downloads the index and representative-cover tables, then asks which larger groups to include. The [dataset reference](../dataset.md) lists these groups.

````python
!plinder_download --help
````

Run this in a terminal to include every group without confirmation:

```bash
plinder_download --yes
```

Rerun the same command after an interruption; completed files are reused. Source PDB mmCIFs are fetched separately when reconstructing structures. For offline reconstruction, first cache the entries you need with `download_pdb_mmcifs()`, described in the {ref}`structure reference <structure-asset-reference>`.

## Work offline

After the files you need are available locally, set this before starting Python:

```bash
export PLINDER_OFFLINE=true
```
