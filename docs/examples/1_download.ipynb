# Downloading the dataset

The default location for the dataset is `~/.local/share/plinder/<PLINDER_RELEASE>/<PLINDER_RELEASE_NUMBER>` where:
- `<PLINDER_RELEASE>` is the month of the PDB snapshot, formatted as `YYYY-MM`;
- `<PLINDER_RELEASE_NUMBER>` distinguishes PLINDER releases from the same month.

Set the `PLINDER_MOUNT` environment variable to use a different location.


## Use `plinder_download` to download the complete dataset

```bash
plinder_download --help
usage:
    Download the full plinder dataset for the current configuration.
    Note that even though this is wrapped in a progress bar, the estimated
    completion time can vary wildly as it iterates over larger files vs.
    smaller ones.
    

optional arguments:
  -h, --help            show this help message and exit
  --release RELEASE     plinder release
  --release-number RELEASE_NUMBER
                        plinder release number
  -y, --yes             skip confirmation
```

:::{note} The public `2024-06/v2` release is served from Cloudflare R2 and contains about 220 GB of downloadable files. Allow additional space for extracted archives; download time depends on your connection. Completed files are reused on subsequent runs. Set `PLINDER_OFFLINE=true` to use the cache without network requests. :::

## Use the `plinder` Python package to lazily access the dataset

Alternatively, if the `PLINDER_OFFLINE` environment variable **unset** (which is the default empty), the dataset will be downloaded on lazily and on the fly as you access the data. This is preferred for exploration and prototyping as you don't need to download the entire dataset at once and can just work with the assets you need for your use-case.

````python
import os

os.environ.setdefault("PLINDER_RELEASE", "2026-07")
os.environ.setdefault("PLINDER_RELEASE_NUMBER", "1")
````

## Access release files from Python

Alternatively, when `PLINDER_OFFLINE` is unset, API calls fetch missing release files as they are requested. This is useful for exploration because only the files needed by the current query or reconstruction are cached.

````python
from plinder.core import get_config

cfg = get_config()
print(f"local directory: {cfg.data.plinder_dir}")
print(f"remote data directory: {cfg.data.plinder_remote}")
````
