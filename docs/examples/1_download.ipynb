# Downloading the dataset

The default location for the dataset is `~/.local/share/plinder/<PLINDER_RELEASE>/<PLINDER_RELEASE_NUMBER>` where:
- `<PLINDER_RELEASE>` is the month of the PDB snapshot, formatted as `YYYY-MM`;
- `<PLINDER_RELEASE_NUMBER>` distinguishes PLINDER releases from the same month.

Set the `PLINDER_MOUNT` environment variable to use a different location.

```python
import os

os.environ.setdefault("PLINDER_RELEASE", "2026-07")
os.environ.setdefault("PLINDER_RELEASE_NUMBER", "1")
```

## Access release files from Python

Alternatively, when `PLINDER_OFFLINE` is unset, API calls fetch missing release files as they are requested. This is useful for exploration because only the files needed by the current query or reconstruction are cached.

```python
from plinder.core import get_config

cfg = get_config()
print(f"local directory: {cfg.data.plinder_dir}")
print(f"remote data directory: {cfg.data.plinder_remote}")
```
